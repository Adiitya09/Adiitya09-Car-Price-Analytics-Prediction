```python
import re

# Extract Torque
df["Torque"] = df["torque2"].str.extract(
    r"(\d+(?:\.\d+)?)"
)

# Extract RPM
df["RPM"] = df["torque2"].str.extract(
    r"([\d,]+)\s*rpm",
    flags=re.IGNORECASE
)
```

```python
# Remove the original torque2 column
df.drop(columns=["torque2"], inplace=True)

# Save the updated dataset
#df.to_excel("Car_details_updated.xlsx", index=False)

df.head()
```

```python
df["Brand"] = df["name"].str.split().str[0]

#df.to_excel("Car_details_updated.xlsx", index=False)

df.head()
```

```python
#Replace Null valaues

df["mileage"] = df["mileage"].fillna(df["mileage"].mean().round(2))
print(df["mileage"].isnull().sum())
```

```python
#Replace Engine Null Values
df["engine"]=df["engine"].fillna(df["engine"].mean().round(2))
print(df["engine"].isnull().sum())
```

```python
#Replace Max_power Null Values
df["max_power"] = (
    df["max_power"]
    .astype(str)
    .str.extract(r"(\d+\.?\d*)")[0]
)

df["max_power"] = pd.to_numeric(df["max_power"], errors="coerce")

df["max_power"] = df["max_power"].fillna(df["max_power"].mean().round(2))
df["max_power"] = df["max_power"].astype(float)
print(df["max_power"].isnull().sum())
```

```python
#Replace Null Torque values
#df["Torque"] = df["Torque"].fillna(df["Torque"].mean().round(2))
df["Torque"] = df["Torque"].astype(float)
print(df["Torque"].isnull().sum())
```

```python
#Repalce Null Seat Values
df["seats"] = df["seats"].fillna(df["seats"].mean().round(2))
df["seats"] = df["seats"].astype(float)
print(df["seats"].isnull().sum())
```

```python
#Replace RPM Null Values
df["RPM"] = (
    df["RPM"]
    .astype(str)
    .str.extract(r"(\d+\.?\d*)")[0]
)

df["RPM"] = pd.to_numeric(df["RPM"], errors="coerce")
df["RPM"] = df["RPM"].fillna(df["RPM"].mean().round(2))
df["RPM"] = df["RPM"].astype(float)
print(df["RPM"].isnull().sum())
```

```python
df = df.round(2)
df.head()
```

```python
df['name'] = df['name'].str.strip()
df['fuel'] = df['fuel'].str.strip()
df['seller_type'] = df['seller_type'].str.strip()
df['transmission'] = df['transmission'].str.strip()
df['owner'] = df['owner'].str.strip()
df['Brand'] = df['Brand'].str.strip()

df.head()
```

```python
df.to_excel("Car_details_updated.xlsx", index=False)
```
